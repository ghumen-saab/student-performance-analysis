```python
df["Class"] = df["Class"].str.strip().str.lower()

df["Class"] = df["Class"].replace({
    "11 th": "11th"
})

df["Class"] = df["Class"].str.replace(
    "10th", "10th"
)

df["Class"] = df["Class"].str.title()
```

```python
df["Class"] = df["Class"].str.strip().str.lower()

df["Class"] = df["Class"].replace({
    "10th": "10th",
    "9th": "9th",
    "11 th": "11th",
    "12th": "12th",
    "unknown": "Unknown"
})
```

```python
df[
    (df["Attendance %"] < 0) |
    (df["Attendance %"] > 100)
]
```

```python
df[
    (df["Final Exam Score"] < 0) |
    (df["Final Exam Score"] > 100)
]
```

```python
df.loc[
    (df["Study Hours"] < 0) | 
    (df["Study Hours"] > 24),
    "Study Hours"
] = np.nan

df.loc[
    (df["Attendance %"] < 0) | 
    (df["Attendance %"] > 100),
    "Attendance %"
] = np.nan

df.loc[
    (df["Final Exam Score"] < 0) | 
    (df["Final Exam Score"] > 100),
    "Final Exam Score"
] = np.nan
```

```python
df.isnull().sum()
```

```python
df["Study Hours"] = df["Study Hours"].fillna(
    df["Study Hours"].median()
)

df["Attendance %"] = df["Attendance %"].fillna(
    df["Attendance %"].median()
)

df["Final Exam Score"] = df["Final Exam Score"].fillna(
    df["Final Exam Score"].median()
)
```

```python
df[
    [
        "Study Hours",
        "Attendance %",
        "Assignment Score",
        "Midterm Score",
        "Final Exam Score",
        "Overall Score"
    ]
].describe()
```

```python
subject_performance = df.groupby(
    "Subject"
)["Overall Score"].mean().sort_values(ascending=False)

subject_performance
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

subject_performance.plot(kind="bar")

plt.title("Average Overall Score by Subject")
plt.xlabel("Subject")
plt.ylabel("Average Overall Score")
plt.xticks(rotation=45)

plt.show()
```

```python
class_performance = df.groupby(
    "Class"
)["Overall Score"].mean().sort_values(ascending=False)

class_performance
```

```python
plt.figure(figsize=(8, 5))

class_performance.plot(kind="bar")

plt.title("Average Overall Score by Class")
plt.xlabel("Class")
plt.ylabel("Average Overall Score")
plt.xticks(rotation=0)

plt.show()
```
