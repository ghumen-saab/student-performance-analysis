```python
# Make a copy of the cleaned dataset
clean_df = df.copy()

# Fill missing categorical values
clean_df["Gender"] = clean_df["Gender"].fillna("Unknown")
clean_df["Class"] = clean_df["Class"].fillna("Unknown")

# Fill missing numerical values with the median
numeric_columns = [
    "Attendance %",
    "Assignment Score",
    "Midterm Score",
    "Final Exam Score"
]

for column in numeric_columns:
    clean_df[column] = clean_df[column].fillna(
        clean_df[column].median()
    )

clean_df.head()
```
